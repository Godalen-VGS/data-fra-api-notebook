Et **API** er en adresse på nettet som gir deg data i stedet for en nettside. Mange offentlige etater deler data gratis på denne måten: Meteorologisk institutt, SSB, NASA og mange flere. Med Python kan du hente dataene og regne på dem direkte.

Dataene kommer som regel i formatet **JSON**, som ser nesten ut som Python-lister og -dicts.

## Slik henter du data

Pakken `requests` henter data fra en adresse:

```python
svar = requests.get(adresse, params={...}, headers={...}, timeout=10)
svar.status_code    # 200 betyr OK
data = svar.json()  # gjør JSON om til Python-lister og -dicts
```

- `params` er det du spør om, for eksempel breddegrad og lengdegrad.
- `headers` er informasjon om deg. Meteorologisk institutt krever at du sier hvem du er i `User-Agent`.
- `timeout=10` gjør at programmet gir opp etter 10 sekunder hvis nettet er tregt.

**Vær høflig:** Ikke hent de samme dataene hundre ganger i en løkke. Hent én gang, lagre i en variabel og regn videre på den.

## Eksempel: værvarsel fra yr

Meteorologisk institutt (MET) står bak yr.no og deler værvarselet sitt på [api.met.no](https://api.met.no). Her henter vi varselet for Stavanger.

````python
import requests

adresse = "https://api.met.no/weatherapi/locationforecast/2.0/compact"
sted = {"lat": 58.97, "lon": 5.73}
hvem = {"User-Agent": "skoleoppgave-python/1.0"}

svar = requests.get(adresse, params=sted, headers=hvem, timeout=10)
print("Statuskode:", svar.status_code)

varsel = svar.json()
forste = varsel["properties"]["timeseries"][0]
print(forste["time"], forste["data"]["instant"]["details"])
````
